```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df=pd.read_csv("/content/amazon.csv")
df
```

```python
df.isnull()
```

```python
df.info()
```

```python
df.drop_duplicates()
```

```python
df.describe()
```

```python
df.drop(['rating'],axis=1,inplace=True)
df.drop(['rating_count'],axis=1,inplace=True)
df.drop(['about_product'],axis=1,inplace=True)
df.drop(['user_name'],axis=1,inplace=True)
df.drop(['review_id'],axis=1,inplace=True)
df.drop(['review_title'],axis=1,inplace=True)
df.drop(['review_content'],axis=1,inplace=True)
df.drop(['img_link'],axis=1,inplace=True)
df.drop(['product_link'],axis=1,inplace=True)
```

```python
df
```

```python
df['product_id']=le.fit_transform(df['product_id'])
df['product_id']
```

```python
df['product_name']=le.fit_transform(df['product_name'])
df['product_name']
```

```python
df['category']=le.fit_transform(df['category'])
df['category']
```

```python
df['user_id']=le.fit_transform(df['user_id'])
df['user_id']
```

```python
df['actual_price']=df['actual_price'].replace(r'[^\d.]','',regex=True).astype(float)
df['discounted_price']=df['discounted_price'].replace(r'[^\d.]','',regex=True).astype(float)
df['discount_percentage']=df['discount_percentage'].str.replace('%','',regex=False).astype(float)
```

```python
df
```

```python
x=df.drop(['discounted_price'],axis=1).values
x
```

```python
y=df['discounted_price'].values
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor()
```

```python
model.fit(x_train,y_train)
```

```python
productid=input("enter productid:")
productname=input("enter productname:")
category=input("enter category:")
actualprice=input("enter actual price:")
discountpercentage=input("enter discount percentage:")
userid=input("enter userid:")

abc=[productid,productname,category,actualprice,discountpercentage,userid]


abc_as_numpy_array=np.asarray(abc)
abc_reshaped=abc_as_numpy_array.reshape(1,-1)
prediction=model.predict(abc_reshaped)
print("Total amount after discount:",prediction)
```
